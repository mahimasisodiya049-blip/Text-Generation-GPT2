```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

print("Loading model... please wait.")

# 1. Load the "brain" (model) and the "dictionary" (tokenizer)
model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

def generate_text(prompt, length=100):
    # 2. Turn your words into numbers the AI understands
    inputs = tokenizer.encode(prompt, return_tensors="pt")
    
    # 3. Tell the AI to create new numbers (words)
    outputs = model.generate(
        inputs, 
        max_length=length, 
        do_sample=True, 
        temperature=0.7, 
        top_p=0.9,       
        no_repeat_ngram_size=2
    )
    
    # 4. Turn those numbers back into a human paragraph
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

# 5. Try it out!
user_input = "The future of space travel is"
result = generate_text(user_input)
print("\n--- GENERATED PARAGRAPH ---")
print(result)
```

```python
def generate_coherent_paragraph(topic_prompt):
    # Prepare the input
    inputs = tokenizer.encode(topic_prompt, return_tensors="pt")
    
    # Generate with specific 'rules' for coherence
    output_sequences = model.generate(
        input_ids=inputs,
        max_length=200,          # Length of the paragraph
        temperature=0.8,         # Creativity: 0.7-0.8 is best for coherence
        top_k=50,                # Limits vocabulary to top 50 best words
        top_p=0.95,              # Nucleus sampling: filters out low-probability words
        no_repeat_ngram_size=3,  # Prevents the model from repeating 3-word phrases
        do_sample=True,          # Enables the 'creativity' settings above
        early_stopping=True
    )

    return tokenizer.decode(output_sequences[0], skip_special_tokens=True)

# Test with a specific topic
print(generate_coherent_paragraph("The evolution of sustainable architecture has led to"))
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Create the UI Elements with Styling
title = widgets.HTML("<h2>AI Text Generation Dashboard</h2>")
description = widgets.HTML("<p>Enter a topic or a starting sentence to generate a coherent paragraph.</p>")

prompt_input = widgets.Textarea(
    value='The future of sustainable energy is',
    placeholder='Type your prompt here...',
    description='Prompt:',
    layout=widgets.Layout(width='90%', height='100px'),
    style={'description_width': 'initial'}
)

generate_btn = widgets.Button(
    description='Generate Paragraph',
    button_style='success', # 'success' makes it green
    tooltip='Click to generate text',
    icon='magic' # Adds a little wand icon
)

output_area = widgets.Output(
    layout={'border': '1px solid #444', 'padding': '10px', 'margin': '10px 0'}
)

# 2. Define the Button Logic
def on_button_clicked(b):
    with output_area:
        clear_output()
        print("AI is thinking... please wait...")
        
        # Call your existing generation function
        try:
            result = generate_coherent_paragraph(prompt_input.value)
            clear_output()
            print(f"PROMPT: {prompt_input.value}\n")
            print("-" * 30)
            print(result)
            print("-" * 30)
        except Exception as e:
            print(f"Error: {e}. Did you run the previous code cells?")

# 3. Connect the Button to the Logic
generate_btn.on_click(on_button_clicked)

# 4. Display Everything in a Vertical Box
ui_layout = widgets.VBox([title, description, prompt_input, generate_btn, output_area])
display(ui_layout)
```
